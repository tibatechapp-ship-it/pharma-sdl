## 📚 النظرية

**نموذج الحجرة الواحدة:**

**IV:** $C(t) = Dose/V_d \cdot e^{-kt}$

**Oral:** $C(t) = F\cdot Dose\cdot k_a/(V_d(k_a-k)) \cdot (e^{-kt} - e^{-k_a t})$


```python
def pk_iv(dose, vd, cl, t_max=24, n=1000):
    k = cl / vd
    t = np.linspace(0.01, t_max, n)
    return t, (dose/vd) * np.exp(-k*t)

def pk_oral(dose, vd, cl, ka, f=1.0, t_max=24, n=1000):
    k = cl / vd
    t = np.linspace(0.01, t_max, n)
    return t, (f*dose*ka)/(vd*(ka-k)) * (np.exp(-k*t) - np.exp(-ka*t))

print('✅ دوال PK جاهزة')
```

```python
t_iv, c_iv = pk_iv(dose=1000, vd=70, cl=21, t_max=12)
t_or, c_or = pk_oral(dose=1000, vd=70, cl=21, ka=2.0, f=0.9, t_max=12)
fig, ax = plt.subplots()
ax.plot(t_iv, c_iv, 'b-', linewidth=2.5, label='IV')
ax.plot(t_or, c_or, 'r-', linewidth=2.5, label='Oral')
ax.set_xlabel('Time (h)'); ax.set_ylabel('Conc (mg/L)')
ax.set_title('Paracetamol 1000 mg — IV vs Oral')
ax.legend()
plt.show()
print(f't½ = {0.693*70/21:.2f} h')
```
